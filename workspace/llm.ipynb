```python
line_count, word_count = 0, 0
sample_lines = ''
number_of_samples = 4
with open('TinyStoriesV2-GPT4-train.txt', 'r', encoding='utf-8') as file:
    for i, line in enumerate(file):
        if i < number_of_samples:
            sample_lines = sample_lines + ' ' + line
            word_count+= len(line.split())
        else:
            break
print(f'words: {word_count}')
print(f'Sample Lines {sample_lines}')
```

```python
import tiktoken 
tokenizer = tiktoken.get_encoding("gpt2")
sample_tokens = tokenizer.encode(sample_lines, allowed_special={"<|endoftext|>"})
for i, token in enumerate(sample_tokens):
    decoded = tokenizer.decode([token])
    print(f' Character:Token {decoded}:{token} \n')
```
